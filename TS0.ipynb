```python
import matplotlib.pyplot as plt
import numpy as np

fs = 1000   #frecuencia de muestreo
N = 1000    #cantidad de muestras

def func_sen(vmax, dc, ff, ph, nn, fs):

    xx = vmax*np.sin(2*np.pi*ff*nn/fs + 2*np.pi*ph) + dc
    tt = nn/fs
    
    return(tt, xx)


vmax = 1                #amplitud
dc = 0                  #offset
ff = 1                  #frecuencia de señal
ph = 0                  #fase
nn = np.arange(0, N)    #vector de muestras

tt, xx = func_sen(vmax, dc, ff, ph, nn, fs)

plt.plot(tt, xx)
plt.title(ff)
plt.xlabel('tiempo [s]')
plt.ylabel('Amplitud [V]')
```
